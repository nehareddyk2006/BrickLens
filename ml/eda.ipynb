```python
import pandas as pd

master = pd.read_csv("../datasets/processed/bricklens_featured.csv")

print(master.shape)
```

```python
import plotly.express as px

city_avg = (
    master.groupby("city")["price_lakh"]
    .mean()
    .reset_index()
)

fig = px.bar(
    city_avg,
    x="city",
    y="price_lakh",
    title="Average Property Price by City"
)

fig.show()
```

```python
sample = master.sample(5000)

fig = px.scatter(
    sample,
    x="sqft",
    y="price_lakh",
    color="city",
    title="Price vs Area"
)

fig.show()
```

```python
fig = px.histogram(
    master,
    x="bhk",
    title="BHK Distribution"
)

fig.show()
```

```python
top_localities = (
    master.groupby("location")["price_lakh"]
    .mean()
    .sort_values(ascending=False)
    .head(15)
    .reset_index()
)

fig = px.bar(
    top_localities,
    x="location",
    y="price_lakh",
    title="Top Expensive Localities"
)

fig.show()
```
